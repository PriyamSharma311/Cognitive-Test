```python
import pandas as pd
import os


def generate_prompt(user_input: str) -> str:
    """
    Generates an adaptive prompt based on the unified child profile.
    """

    profile_path = "/Users/priyamsharma/Documents/Et Gen ai hackathon/Data/processed/unified_child_profile.csv"

    if not os.path.exists(profile_path):
        raise FileNotFoundError(
            "unified_child_profile.csv not found. "
            "Run feature_engineering.py first."
        )

    # Load unified profile (single-row state)
    profile = pd.read_csv(profile_path).iloc[0]

    cognitive_level = profile["cognitive_level"]
    autism_risk = profile["autism_risk"]
    dominant_emotion = profile["dominant_emotion"]

    rules = []

    # Cognitive adaptation
    if cognitive_level == "low":
        rules.append("Use very simple words and short sentences.")
    elif cognitive_level == "medium":
        rules.append("Use simple explanations with examples.")
    else:
        rules.append("Use normal conversational language.")

    # Autism-specific adaptation
    if autism_risk == "high":
        rules.append("Use structured steps and repeat key instructions.")
    elif autism_risk == "medium":
        rules.append("Keep responses predictable and organized.")

    # Emotion-aware adaptation
    if dominant_emotion in ["sad", "angry", "fear"]:
        rules.append("Use a calm, empathetic, and reassuring tone.")
    elif dominant_emotion == "happy":
        rules.append("Use an encouraging and positive tone.")

    # Build final prompt
    prompt = f"""
You are a supportive AI companion for children with learning
and developmental challenges.

Communication Rules:
{chr(10).join("- " + rule for rule in rules)}

Child says:
{user_input}

Respond kindly and clearly following the rules above.
"""

    return prompt


# Quick test
if __name__ == "__main__":
    print(generate_prompt("I don't understand my homework"))
```
